# 上市公司基本資料

```python
#【上市公司基本資料】
#https://data.gov.tw/dataset/18419 
url='https://mopsfin.twse.com.tw/opendata/t187ap03_L.csv'
headers = {
    "User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/85.0.4183.102 Safari/537.36"
}
req = requests.get(url, headers = headers,verify=False)
data = req.content.decode('utf8')
df_listing = pd.DataFrame([x.replace('"','').split(',') for x in data.split('\n')])
df_listing  = pd.DataFrame(df_listing .values[1:], columns=df_listing .iloc[0])
df_listing['Type']='上市'
df_listing1=df_listing[['公司代號', '公司名稱', '公司簡稱','產業別','實收資本額','Type']]
#【解碼產業】
#【Data 合併】
df_listing2=pd.merge(df_listing1, mappingData, how='left', left_on='產業別', right_on='新編碼')
```

# 上櫃公司基本資料

```python
# UI:https://data.gov.tw/dataset/25036
url='https://mopsfin.twse.com.tw/opendata/t187ap03_O.csv'
#若有 CSV 可以選用True 做更新 False 會直接讀2021/07/16 下載的資料
updatamappingData=False
headers = {
    "User-Agent":"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/85.0.4183.102 Safari/537.36"
}
req = requests.get(url, headers = headers,verify=False)
data = req.content.decode('utf8')
dfOCT = pd.DataFrame([x.replace('"','').split(',') for x in data.split('\n')])
dfOCT  = pd.DataFrame(dfOCT .values[1:], columns=dfOCT .iloc[0])
dfOCT['Type']='上櫃'
dfOCT1=dfOCT[['公司代號', '公司名稱', '公司簡稱','產業別','實收資本額','Type']]

#【解碼產業】
#【Data 合併】
dfOCT_=pd.merge(dfOCT1, mappingData, how='left', left_on='產業別', right_on='新編碼')
```
